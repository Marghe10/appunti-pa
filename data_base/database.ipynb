Paprire una connessione puo comportare tutta una serie di errori, quindi normalmente si mette il codice che apre una connesione dentro un blocco try/except. I comandi che uso per definire gli errori andranno importati, ad esempio mysql.errorcode sarà sottolineato in rosso e ti dirà che va importato

```python
from errno import errorcode
import mysql.connector

try:
    cnx=mysql.connector.connect(user='root', password='', host='localhost', port="3306", database='products_suppliers')

except mysql.connector.Error as err:
    if err.errno == errorcode.ER_ACCESS_DENIED_ERROR:
        print("ERRORE: Qualcosa è andato storto con username o password")
    elif err.errno == errorcode.ER_BAD_DB_ERROR:
        print("ERRORE: Il Database non esite")
    else:
        print(err)
else:
    cnx.close()
```

CURSORE

Il cursore è un "affare" che si trova all'inizio della tabella e che puoi spostare sulle varie righe. Può andare avanti una riga alla volta o saltare un certo numero di righe o andare in fondo ecc... Un po' come sui file una volta che ho letto tutto devo riportare il cursore all'inizio.

```python
import mysql.connector

try:
    cnx=mysql.connector.connect(user='root', password='', host='localhost', port="3306", database='products_suppliers')
    cursore=cnx.cursor() #creo il cursore


except mysql.connector.Error as err: # Per vedere bene come gestire gli errori prendi il codice sopra
    print(err)
else:
    cnx.close()
```

Il cursore è un modo con cui io posso eseguire una interrogazione al database. Ci sono diversi tipi di cursore:

Quello più semplice si chiama semplicemente "Cursor"

Una volta aperto il cursre posso scrivere il mio codice SQL e poi devo eseguirlo usando il cursore.

```python
import mysql.connector

try:
    cnx=mysql.connector.connect(user='root', password='', host='localhost', port="3306", database='products_suppliers')
    cursore=cnx.cursor()
    query="select * from S"
    cursore.execute(query)

except mysql.connector.Error as err: # Per vedere bene come gestire gli errori prendi il codice sopra
    print(err)
else:
    cnx.close()
```

IL CURSORE è CONTENITORE DI RIGHE

Il cursore è un contenitore di righe della tabella, per cui posso iterare al suo interno come in un qualsiasi altro contenitore. Uso for

```python
import mysql.connector

try:
    cnx=mysql.connector.connect(user='root', password='', host='localhost', port="3306", database='products_suppliers')
    cursore=cnx.cursor()
    query="select * from S"
    cursore.execute(query)

    for riga in cursore:
        print(riga)
        # posso anche scrivere cose tipo print(riga[1]) e mi stampa tutti i nomi che sono contenuti nella casella 1 delle varie righe della tabella S

except mysql.connector.Error as err: # Per vedere bene come gestire gli errori prendi il codice sopra
    print(err)
else:
    cnx.close()
```

CURSORI DI VARIO TIPO --> CURSORE DIZIONARIO



```python
import mysql.connector

try:
    cnx=mysql.connector.connect(user='root', password='', host='localhost', port="3306", database='products_suppliers')
    cursore_dizionario=cnx.cursor(dictionary=True)
    query="select * from S"
    cursore_dizionario.execute(query)

    for riga in cursore_dizionario:
        print(riga)

except mysql.connector.Error as err: # Per vedere bene come gestire gli errori prendi il codice sopra
    print(err)
else:
    cnx.close()
```

In pratica fa lo stesso che il connettore normale ma con la differenza che gli attributi che prende dal database gli inserisce in una lista di dizionari che hanno la prima colonna del database (in questo caso i codici) come chiave. Ora quindi posso stampare gli elementi che mi servono senza usare (riga[1]), ma usando i nomi degli attributi con davanti il nome della casella, come abbiamo visto prima NAME --> SNAME. Quindi posso scrivere (riga["SNAME"]).

```python
import mysql.connector

try:
    cnx=mysql.connector.connect(user='root', password='', host='localhost', port="3306", database='products_suppliers')
    cursore_dizionario=cnx.cursor(dictionary=True)
    query="select * from S"
    cursore_dizionario.execute(query)

    for riga in cursore_dizionario:
        print(riga["SNAME"])

except mysql.connector.Error as err: # Per vedere bene come gestire gli errori prendi il codice sopra
    print(err)
else:
    cnx.close()
```
