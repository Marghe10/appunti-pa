Come importare il database da 0

Dopo aver installato Xamp e DBeaver apri Xamp e avvia MySql. Poi apri DBeaaver e in altro a sinistra premi sull'icona della presa con un piccolo "+" vicino. Premi sopra e fai doppio clic su MySQL. Ora che si è creato il collegamento e nella colonna a sinistra vedi scritto "local host 3306" fai tasto destro sulla scritta local host. Si apre una tendina, la prima voce è SQL editor, premi e poi premi su "new SQL script". A queso punto nella barra in altro vedi la voce "SQL editor", premi sopra e si apre un menù a tendina. In basso vedi scritto "import SQL script", ci premi sopra e selezioni il file da dentro le cartelle del tuo computer. A questo punto posso eseguire lo script con il comando a forma di pergamena che si trova nella colonnina verticale sulla sinistra e che dice "execute SQL script".

Ora devo fare tasto destro sulla scritta database nella colonna a sinistra e premere "refresh" in modo che appaia il database che ho appena creato
