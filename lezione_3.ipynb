ECCEZIONI

Le eccezioni sono un modo per gestire gli errori.

Per gestire gli errori si inseriscono i blocchi di codice in un "Try" in modo da poter inserire delle eccezioni quando necessario usando il comando "except".

```python
input_sbagliato = True
while input_sbagliato==True:
    stringa=input("inserisci un numero intero")
    try:
        numero=int(stringa)
        risultato=numero/2
        print(f"il risultato è: {risultato}")
        input_sbagliato=False
    except ValueError:
        print("ho detto di inserire un numero intero!!")
    except ZeroDivisionError:
        print("non è possibile dividere per zero")
    except TypeError:
        print("qualcosa è andato storto...")
```

Cosi io entro nel while e inserisco qualcosa nell'input. Se l'input è giusto allora la funzione int lo trasforma in intero, poi viene diviso per due, poi viene stampato il risultato e poi si trasforma la variabile input_sbagliato in False, quindi quando il puntatore risale al while per eseguirlo si accorge che while input_sbagliato==True non va più eseguito perchè input_sbagliato ora è == False.
Se invece inserisco un valore che non è un numero intero, quando il mio valore entra nella funzione input la funzione si rompe e scatena un errore che è diverso a seconda di quello che ho inserito, il puntatore entra nel except corretto, stampa il messaggio di errore e poi torna al while e lo riesegue perchè input_sbagliato è ancora == True.

COPIA UNA LISTA

```python
luca=[11,22, 33]
mannella=luca
marghe=list(luca)

marghe[1] = 7
print(luca)
print(mannella)
print(marghe)
```

Io ho una lista di 3 numeri, sia luca che mannella sono puntatori che vanno a leggere o eventualmente modificare quella lista. Quindi se io provo a modificare luca facendo tipo luca[1]=0 e poi faccio pint(mannella), l'output del codice sarà [11,0,33] ovvero la mia lista modificata usando luca. Perchè luca e mannella sono semplicemente due modi di indicare la stessa lista.

Marghe invece non è un puntatore alla lista luca/mannella, ma è una copia di quella lista. il comando list(luca) ha copiato gli elementi che erano dentro la lista luca in un'altra lista che ho chiamato marghe, quindi poi se combio degli elementi in marghe quegli elementi non saranno modificati in luca, per cui se scrivo marghe[0]= 5 e poi stampo marghe e luca il mio output sarà: marghe=[5,22,33] e luca=[11,22,33] quindi solo marghe è stato modificato

APPEND E INSERT


```python
lista=[11,22,33,44,55,66,77]
print(f"la lista prima delle modifiche è: {lista}")
lista.append(100)
print(f"la lista dopo  l'append 100 è: {lista}")
lista.insert(2,0)
print(f"la lista dopo insert 0 è: {lista}")
```

.append(1) aggiunge alla lista il valore 1 in ultima posizione. Se vuoi scegliere in che posizione inserire una nuova variabile devi usare .insert(index,object) e inserire prima in quale posizione vuoi inserire la nuova variabile epoi quale vuoi che sia la nuova variabile. Se invece vuoi cambiare una variabile non si fa con le funzioni della lista.

```python
lista=[11,22,33,44,55,66,77]
lista[3]=100
print(lista)
```

hai sostituito l'elemento tre della lista con il numero 100

JOIN()

In Python, join() serve a unire più stringhe in una sola, inserendo tra loro un separatore scelto da te.

```python
lista_parole = ["Ciao", "come", "stai"]
frase1 = " ".join(lista_parole) #in questo caso sto mettendo uno spazio come separatore
frase2= ",".join(lista_parole)
frase3= "-".join(lista_parole)

print(f"la prima frase è: {frase1}\nla seconda frase è: {frase2}\nla terza frase è: {frase3}")
```

SPLIT()

split() fa, in pratica, l’operazione opposta a join(): divide una stringa in una lista di stringhe. Se non specifichi nulla, split() separa in corrispondenza degli spazi bianchi o di altri caratteri bianchi, come il fine riga \n. Puoi anche indicare un separatore preciso, per esempio la virgola.
split() è un modo per tagliare una stringa in pezzi: gli dici dove separarla e Python ti restituisce una lista di stringhe.

```python
frase = "Ciao come stai"
lista_parole = frase.split()
print(lista_parole)

dati = "Mario,Luisa,Anna"
lista_nomi = dati.split(",")
print(lista_nomi)  # ['Mario', 'Luisa', 'Anna']
```

```python
file=open("poesia_2","r")
for line in file:
    word_list=line.split()
    print(word_list)
    for i in word_list: #"i" passa tutte le parole della poesia una alla volta
        word=i.rstrip(".,!?") # la variabile word corrisponde ad ogni iterazione ad una parola diversa e il rstrip serve a                               togliere alla parola eventuale punteggiatura aggiuntiva.
        print(word)
file.close()
```
