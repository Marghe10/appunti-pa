Passo da seguire per il print. Di default il passo che segue la funzione while segue è il passo uno, quindi itera ad ogni lettera o ad ogni numero.

```python
nome = "MArguerite"
print(nome[::2]) # così l'iterazione è ogni due lettere partendo da 0 e andando fino alla fine della                         parola. Equivale a scrivere: print(nome[0:11:2]) riordando che Marguerite ha 10 lettere                   e che 0:10 stampa solo dalla 0 alla 9.
```

posso applicare il passo anche alla lettura di un vettore. Ricordo che il vettore si riempie scrivendo nome del vettore.append() e inserendo tra parentesi che cosa si desidera aggiungere.

```python
import random

i = 0
numero = []
while i < 100:
    numero.append(random.randint(0, 100))
    i += 1
print(numero[::2])
```

ENUMERATE

E uno strano comando che prende in input una stringa a restituisce l'index piu il valore espresso in quella posizione.
Se gli dai in pasto la parola "HELLO", restituisce la lettera H indicando l'index 0, poi la E con index 1, poi la L con index 2 ecc...
Se vuoi associarlo ad una variabile devi usare un ciclo for che scorre su tutte le variabili che restituisce enumerate, non puoi fare un'associazione classica con x=enumerate(stringa) perchè enumerate produce una coppia di valori per ogni carattre presente nella stringa, non posso inserirli tutti in una sola x.

```python
stringa="la vita è bella"
enumerate(stringa)
for (index,letter) in enumerate(stringa):
    print(f"{index} : {letter}")
```

Un altro modo per printare il risultato è infilare "enumerate()" in una lista. Questo è un buon metodo se ho poi bisogno di mettere mano ai valori che enumerate() tira fuori, perche inserendoli in una lista posso modificarli piu avanti.

```python
stringa = "la vita è bella"
elementi = list(enumerate(stringa))
print(elementi)
```

Così posso poi lavorare sui silgoli elementi

```python
stringa = "la vita è bella"
elementi = list(enumerate(stringa))
index,lettera=elementi[3]
print(f"l'indice è: {index} la lettera è: {lettera}")
```
