I FILE CSV

I fili csv (comma separated value) sono file che contengono campi separati da virgole. Se i campi non contengono punteggiatura ma sono semplicemente frasi o parole allora il csv è un file con parole e frasi divise da virgole. Se invece voglio inderire delle virgole all'internno di un campo devo delimitare i campi con delle virgolette.

libro, autore, anno, casa editrice, nome, cognome, protagonista, luogo di nascita, colore. Questo può essere il contenuto di un file csv.

"detective story","1953","airport 1955","Sam,William". Questo è un altro esempio di csv che contine virgolette perchè sennò "Sam, William" sarebbe stato diviso in due campi.

```python
file=open("file_csv.txt","r")
for line in file:
    line_split=line.split(",")
    for word in line_split:
        word=word.strip().strip('"')
        print(word)
```

Ricordo che questo funziona solo se all'interno dei campi non ci sono virgole, perchè il comando line.split() divide ovunque dove trova una virgola, anche se questa è all'interno di un campo delimitato da virgolette.//In word.strip().strip('""') il primo strip() serve ad eliminare spazi e \n ai bordi; il secondo .strip('"') elimina le virgolette ai bordi. Non tocca le virgolette dentro una parola.

Questo pezzo di codice mi permette di leggere il file e dividerlo in un elenco di campi eliminando nell'utput le virgole che dividono le parole nel file_csv.txt e le virgolette.

Ma dato che i file CSV sono molto importani python ha un comando apposta per leggerli: reader()

```python
from csv import reader
file=open("file_csv.txt","r") # il file va cq aperto prima in read, dopo posso usare reader
csvReader = reader(file)
for row in csvReader :
    print(row)
```

csvReader è diventato un contenitore di tutto il testo e poi "row" è diventato un contenitore delle parole che compongono le righe di csvReader. Quindi se voglio estrarre il primo campo (tutti i titoli delle opere) posso usare il puntatore dentro un contenitore. Ovvero posso scrivere row[0]

```python
from csv import reader
file=open("file_csv.txt","r")
csvReader = reader(file)
for row in csvReader:
    campo1=row[0]
    print(campo1)
```
