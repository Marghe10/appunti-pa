LAVORARE COI FILE

Per aprire un file è necessario inserire nella funzione "open()" il nome del file e il modo in cui lo si desidera aprire: il readin o writing


```python
filein=open("primo_file","r")
```

Quando apri il file la funzione open() mette a disposizione una sorta di cursore che si posiziona all'inizio del file e ogni volta che voi chimate la funzione read() lui avanza. Avanza a leggere di quanto tu gli dici di avanzare.
Ad esempio:
prima_letera=filein.read(3)
 print(prima_lettera) --> leggerrà e stamperà le prime 3 lettere contenuta nel mio file

```python
filein=open("primo_file","r")
prima_lettera=filein.read(3)
print(prima_lettera)
```

Se voglio leggere tutto il file posso ad esempio usare un ciclo while spendo che quando la funzione read() restituisce una stringa vuota "" quando arriva alla fine del testo. Quindi posso scrivere un pezzo di codice che sia tipo while il mio valore è diverso da "" allora leggi.
Ricorda sempre di chiudere il file dopo averci lavorato sopra perchè sennò alla prossima operazione che ci vuoi fare sopra il cursore sarà dove lo hai lasciato all'operazione prima, in questo caso, visto che ho letto tutto il file, il mio curdore si trova a fondo pagina, quindi se provo a leggere qualcosa senza chiudere prima il file o riportare prima il cursore all'inizio non leggerà niente.

```python
filein=open("primo_file","r")
lettera=filein.read(1)
while lettera != "":
    print(lettera, end="") #aggiungo end=" " cosi non va a capo ogni volta
    lettera=filein.read(1) #questo 1 significa che deve leggere una lettera alla volta
filein.close()
```

```python
filein=open("primo_file","r")
riga=filein.readline()
while riga != "":
    print(riga,end="") # end è necessario perchè nel mio file è già presente un \n sottointeso alla fine di ogni frase, per                       cui se unisco il \n già presente nel file più quello che viene automaticamente inserito dopo il print                    mi vengono o2 \n di fila e quindi nel mio output avro un a capo e poi un altro a capo e quindi una                       riga vuota tra una riga di testo e l'altra.
    riga=filein.readline()
filein.close()
```

Un modo piu furbo per stampare un testo intero è usare la funzione for che scorre sulle righe, perchè anche il file è un contenitore di righe e lettere.

```python
filein=open("primo_file","r")
for i in filein:
    print(i, end="")
filein.close()
```

CONVERTIRE CONTENUTO FILE DA STRINGE A NUMERI: rstrip()

Anche se nel file hai scritto numeri, Python li legge inizialmente come testo.
Supponiamo che il file contenga:
12
2.5
Se scrivi:

```python
with open("file_numeri_2", "r") as file:
    riga1 = file.readline()
    riga2 = file.readline()
print(f"La riga 1 contiene: {riga1} ed è una variabile di tipo: {type(riga1)}")
print(f"La riga 2 contiene: {riga2} ed è una variabile di tipo: {type(riga2)}")
file.close()
```

Otterrai due valori di tipo stringa "12" e "2.5".
Puoi convertirli usando:

```python
with open("file_numeri_2", "r") as file:
    riga1 = file.readline()
    riga2 = file.readline()
numero_intero = int(riga1)
numero_decimale = float(riga2)
print(numero_intero,numero_decimale)
print(f"{type(numero_intero)},{type(numero_decimale)}")
```

int() converte in intero, mentre float() converte in numero decimale. Python riesce spesso a ignorare automaticamente il carattere di fine riga quando usi int() o float(), mentre quando leggi una riga con readline(), normalmente viene incluso anche il carattere speciale "\n".
Per esmepio la prima riga del file potrebe essere: riga1="12\n".
Anche se stampandola normalmente vedi soltanto 12

Per controllare il contenuto reale di una stringa puoi usare il comando repr() che ti mostra anche i caratteri speciali \n \d ecc...

```python
print(repr(riga1))
```

Il problema dei caratteri nascosti è che potrebbero creare problemi soprattutto quando vuoi confrontare le stringhe.

```python
riga = file.readline() #immaginiamo che nella prima riga del file ci sia scritto "ciao" e poi un a capo
if riga == "ciao":
    print("Trovata")
```

Potrebbe non funzionare perché la riga contiene in realtà "ciao\n".

Per eliminare il fine riga e gli eventuali spazi alla fine puoi usare: riga = riga.rstrip()

```python
riga = "ciao\n"
riga = riga.rstrip()
print(repr(riga))
# 'ciao'
```

Senza argomenti, rstrip() rimuove gli spazi bianchi finali, compresi: \n, \t e spazi bianchi normali. La documentazione Python specifica che rstrip() restituisce una copia della stringa senza i caratteri bianchi alla fine.

FUNZIONE SPLIT

split() è un modo per tagliare una stringa in pezzi: gli dici dove separarla e Python ti restituisce una lista di stringhe.

```python
frase = "la vita è bella"
parole = frase.split()
print(parole)
```

Senza argomenti, split() separa dove trova spazi o altri caratteri bianchi, come il fine riga \n. Se ci sono più spazi consecutivi, li tratta come un’unica separazione.

```python
file=open("poesia_2","r")
for line in file:
    word_list=line.split()
    print(word_list)
    for i in word_list: #"i" passa tutte le parole della poesia una alla volta
        word=i.rstrip(".,!?") # la variabile word corrisponde ad ogni iterazione ad una parola diversa e il rstrip serve a                               togliere alla parola eventuale punteggiatura aggiuntiva.
        print(word)
file.close()
```
