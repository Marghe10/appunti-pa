I CONTROLLI

Nella pagina posso inserire i "controlli", ovvero degli elementi grafici. Per vedere quali posso essere gli elementi grafici puoi controllare l'elenco alla pagina: "https://flet.dev/docs/controls". Equivale ad antrare nel sito: "https://flet.dev/docs/" e nel menù di sinistra andare su reverence e poi controls.

Come si fa ad aggiungere un controllo?

Tutti gli elementi grafici si trovano all'interno del modulo ft. Ti metti a scrivere dentro il "ef main" e scrivi "ft.". Vedrai all'interno tutte le classi che puoi utilizzare.

TEXT

un esempio è il comando text: ft,text. Per vedere come usarlo basta andare sui comandi flet al link indicato sopra e inserire text nella barra di ricerca. Poi premi sulla casella text e ti mostra come deve essere formulato il comando:

```python
# Ho trovato il comando text sul sito di flet e chedo che scrive questo:
ft.Text("Qui puoi inserire il testo che desideri", size=24, weight=ft.FontWeight.W_600)
#mi mostra tutto quello che posso inserire nel modulo di tipo testo
```

poi devo assegnare il comando ad una variabile e appendere la variabile alla pagina vuota. Infine devo updateare la pagina. Vedi sotto:

```python
import flet as ft

def main(page:ft.Page):
    il_mio_testo=ft.Text("questo è il mio primo testo!!!", size=24, weight=ft.FontWeight.W_600)
    # ho scritto il testo, poi indicato la dimensione, il tipo e il colore
    # per cambiare il colore posso anche scrivere color="blue"
    page.controls.append(il_mio_testo)
    page.update()
ft.run(main)
```

Vedrai che sul sito, sotto la parte dove ti fa vedere come si costruisce la casella di testo potrai vedere tutti gli aggettivi che puoi aggiungere.

Se voglio cambiare il contenuto della scritta "questo è il mio primo testo!!!" devi scrivere il_mio_testo.value="nuovo testo". La parte della scritta corrisponde al campo "value". Poi devi updatare a pagina

```python
import flet as ft

def main(page:ft.Page):
    il_mio_testo=ft.Text("questo è il mio primo testo!!!", size=24, weight=ft.FontWeight.W_600)
    page.controls.append(il_mio_testo)
    page.update()

    il_mio_testo.value="ciao, sto cambiando il testo"
    page.update()
ft.run(main)
```
