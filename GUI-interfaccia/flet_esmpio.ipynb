AWAIT

è un comando che serve a dire alla pagina flet di aspettare un tot di tempo prima di aggiornarsi

```python
import asyncio
import flet as ft

async def main(page:ft.Page):
    il_mio_testo=ft.Text("questo è il mio primo testo!!!", size=24, weight=ft.FontWeight.W_600)
    page.controls.append(il_mio_testo)
    page.update()

    for i in range (1,100):
        il_mio_testo.value=f"{i}"
        il_mio_testo.update()
        await asyncio.sleep(1)
        page.update()
```

Per poter usare il metodo "await" bisogna che davanti al def venga inserita la parola "async", mentre per usare "asyncio.sleep" bisogna importare asyncio.

Una volta impostata ina scritta decido di sostituirla con un numeo che si itera da 1 a 100. Creo un ciclo for e aggiorno il parametro "value" usando "i" che è il mio puntatore che a ogni iterazione del ciclo for sale di un numero. Immaginiamo di essere alla prima iterazione, "i" vale 1, "il_mio_testo.value" viene aggiornato e diventa 1. Poi update il valore del testo, poi aspetta un secondo, poi aggiorna la pagina.

Così aggiungi anche un bottone (il CupertinoButton è semplicemente un bottone senza bordi tipo di apple che ha sede a cupertino in Silicon valley).

Il problema è che così il bottone viene messo a capi rispetto alla casella di testo, per metterle sulla stessa riga devo creare una riga, inserire gli elementi che voglio mettere in quella riga nell'ordine in cui li voglio e poi appendere la riga alla pagina. Quando crei una riga puoi usare il comando "alignement" per posizionare gli elementi della tua iga dove vuoi.

```python
import flet as ft

def main(page: ft.Page):
    myText = ft.TextField(label="aggiungi elemento:")
    myAddButton = ft.CupertinoButton(content="aggiungi")

    row1=ft.Row([myText, myAddButton], alignment=ft.MainAxisAlignment.CENTER)

    page.controls.append(row1)


ft.run(main)
```

Ora devo inserire le istruzioni che descrivono cosa deve succedere nel momento in cui premo il pulsante "aggiungi". In particolare devo aggiungere delle check box dove posso segnare quando ho comprato quell'oggetto.

```python
import flet as ft

def main(page: ft.Page):
    myText = ft.TextField(label="aggiungi elemento:") # value è il modo in cui si chiama il contenuto della casella di testo, label è solo l'etichetta che vedi prima di scrivere

    def gestione_bottone():
        checkbox=ft.Checkbox(myText.value)
        page.controls.append(checkbox)
        page.update()

    myAddButton = ft.CupertinoButton(content="aggiungi", on_click=gestione_bottone)
    row1=ft.Row([myText, myAddButton], alignment=ft.MainAxisAlignment.CENTER)
    page.controls.append(row1)

ft.run(main)
```

Ora l'ultimo problema da risolvere è che una volta creati piu elementi di quelli che la pagina puo contenere io non posso scorrere verso il basso per vedere quali elementi vengono aggiunti. Quindi invece che appendere gli elementi alla pagina li devo appendere a una lista scorrevole che è un contenitore che puo contenere altre informazioni. In particolare la "listview" è la lista scorrevole alla quale devo aggiungere gli elementi e poi devo aggiungere la list view lla pagina.

```python
import flet as ft

def main(page: ft.Page):
    myText = ft.TextField(label="aggiungi elemento:")
    myListView = ft.ListView(expand=True)
    def gestione_bottone():
        checkbox=ft.Checkbox(myText.value) # value è il modo in cui si chiama il contenuto della casella di testo, label è solo l'etichetta che vedi prima di scrivere
        myListView.controls.append(checkbox)
        page.update()

    myAddButton = ft.CupertinoButton(content="aggiungi", on_click=gestione_bottone)

    row1=ft.Row([myText, myAddButton], alignment=ft.MainAxisAlignment.CENTER)

    page.controls.append(row1)
    page.controls.append(myListView)


ft.run(main)
```
