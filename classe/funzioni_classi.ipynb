INIT

Una classe in Python è un modello con cui puoi creare oggetti che raccolgono insieme dati e operazioni su quei dati. I dati si chiamano attributi; le operazioni si chiamano metodi. Vediamo come si definisce una classe:



```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
```

La funzione __init__ è un metodo speciale che python richiama per inizializzare i dati. è sempre necessario inserire self come primo elemento perchè è l'oggetto che stai inizializzando (si fa così e basta). Tutti i parametri (meno self) vanno messi all'interno di un attributo che si definisce con self. e poi il nome dell'attributo. Il metodo init viene chiamato metodo costruttore

METODO REPR

Il metodo __repr__ serve a definire una rappresentazione testuale di un oggetto: in pratica, dici a Python «quando vuoi mostrare questo oggetto in modo tecnico e informativo, usa questa stringa». È utile soprattutto per capire quali dati contiene un oggetto mentre programmi o cerchi un errore.



```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
    def __repr__(self):
        return f'{self.__class__.__name__} (titolo: {self.titolo}, artista: {self.artista}, anno: {self.anno}, durata: {self.durata}'

#la classe sopra definita dovrebbe essere contenuta in un altro file mentre ciò che c'è scritto di seguito fa parte del file main()

canzone=Canzone("someone like you", "Adele", "2011", "3.11")
print(canzone)
```

{self.__class__.__name__} serve a stapare "Canzone" prima della descrizione dei vari attributi

Tutte le canzoni ce vengono stampate anche nel main chiamando la classe Canzone che contiene il metodo repr saranno printate con la struttura definita nel repr.

Per fare in modo di poter confrontare questi due oggetti devo definire nella classe il metodo __eq__

```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
    def __eq__(self, other):
        if self.titolo==other.titolo:
            return True
        elif self.artista==other.artista:
            return True
        else:
            return False
```

METODO LT

Il metodo LT è il metodo che si implementa per ordinare gli oggetti di una classe (lt sta per lower than). Serve per dire secondo quale parametro devi ordinare quella lista di oggetti

```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
    def __repr__(self):
        return f'{self.__class__.__name__} (titolo: {self.titolo}, artista: {self.artista}, anno: {self.anno}, durata: {self.durata}'
    def __lt__(self, other):
        return self.anno<other.anno #facendo cosi ordino in modo crescente, se volessi ordine decrescente dovrei usare >

# la classe sopra definita dovrebbe essere contenuta in un altro file mentre ciò che c'è scritto di seguito fa parte del file main()

c1= Canzone("someone like you", "Adele", "2011", "3.11")
c2= Canzone("river", "Marghe", "2009", "4.11")
c3= Canzone("someone like you", "Adele","2010", "3.10")
lista_canzoni=[c1,c2,c3]
print("lista canzoni originale")
for c in lista_canzoni:
    print(c)

print("lista canzoni ordinata per anno")
lista_ordinata = sorted(lista_canzoni)
for c in lista_ordinata:
    print(c)

```

Immagino di voler ordinare la lista per anno e a parità di anno ordinare per lunghezza:

```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
    def __repr__(self):
        return f'{self.__class__.__name__} (titolo: {self.titolo}, artista: {self.artista}, anno: {self.anno}, durata: {self.durata}'
    def __lt__(self, other):
        if self.anno == other.anno:
            return self.durata<other.durata
        else:
            return self.anno<other.anno

# la classe sopra definita dovrebbe essere contenuta in un altro file mentre ciò che c'è scritto di seguito fa parte del file main()

c1= Canzone("someone like you", "Adele", "2011", "3.11")
c2= Canzone("river", "Marghe", "2009", "4.11")
c3= Canzone("someone like you", "Adele","2010", "3.10")
c4= Canzone("someone you loved", "Adele","2010", "3.20")
lista_canzoni=[c1,c2,c3,c4]

print("lista canzoni originale")
for c in lista_canzoni:
    print(c)

print("lista canzoni ordinata per anno")
lista_ordinata = sorted(lista_canzoni)
for c in lista_ordinata:
    print(c)
```

METODO KEY

i metodo key si puo usare per ordinare gli oggetti in modo più compatto.

```python
from operator import attrgetter

class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
    def __repr__(self):
        return f'{self.__class__.__name__} titolo: {self.titolo}, artista: {self.artista}, anno: {self.anno}, durata: {self.durata}'
    def __lt__(self, other):
        if self.anno == other.anno:
            return self.durata<other.durata
        else:
            return self.anno<other.anno

# la classe sopra definita dovrebbe essere contenuta in un altro file mentre ciò che c'è scritto di seguito fa parte del file main()

c1= Canzone("someone like you", "Adele", "2011", "3.11")
c2= Canzone("river", "Marghe", "2009", "4.11")
c3= Canzone("someone like you", "Adele","2010", "3.10")
c4= Canzone("someone you loved", "Adele","2010", "3.20")
lista_canzoni=[c1,c2,c3,c4]

print("lista canzoni originale")
for c in lista_canzoni:
    print(c)

print("lista canzoni ordinata per titolo")
#Ci sono due modi di usare key.
# Primo modo:
# lista_ordinata=sorted(lista_canzoni, key=lambda c: c.titolo)
# secondo modo:
lista_ordinata = sorted(lista_canzoni, key=attrgetter('titolo')) #ricorda che per usare questo metodo devi importare attrgetter da perator
for c in lista_ordinata:
    print(c)
```

se voglio ordinare gli oggetti senza aver definito il metodo lt o comunque li voglio ordinare in modo diverso da come avevo definito prima allora posso usare key

DATACLASS

è un decoratore che permette di creare una classe senza bisogno di definire tutti i metodi.

```python
from dataclasses import dataclass,field

@dataclass (eq=True, order=True)
class Canzone:
    titolo : str
    artista : str
    anno : int = field(compare=False)
    durata : str # lo metto come stringa perchè ho un valore scritto "minuti:secondi"

c = Canzone("T", "A", "1999", "5:23")
print(c)


c1= Canzone("someone like you", "Adele", "2011", "3.11")
c2= Canzone("someone like you", "Adele", "2009", "3.11")
if c1==c2:
    print("le canzoni sono uguali")
```

per poter usare il dataclass devi scrivere all'inizio @dataclass che devi aver imortato da tadaclasses. Poi crei la classe e scrivi per ogni attributo di che tipo è. Se costruisci l'oggetto canzone attribuendo così come vedi sopra e poi lo printi vedrai che l'output esce gia ordinato come se avessi implementato un metodo repr.

La dataclass crra anche un metodo di confronto che usa tutti gli attributi, ma non crea un metodo di ordinamento (lt). Se vogliamo dire alla dataclass cosa andare a creare lo dobbiamo indicare vicino a @dataclass.

Se vogliamo che uno dei metodi ignori un determinato attributo bisogna scrivere vicino a quel attributo quale metodo non lo devo considerare usando l'annotazione "= field (campare=false)".
