INIT

Una classe in Python è un modello con cui puoi creare oggetti che raccolgono insieme dati e operazioni su quei dati. I dati si chiamano attributi; le operazioni si chiamano metodi. Vediamo come si definisce una classe:



```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
```

La funzione __init__ è un metodo speciale che python richiama per inizializzare i dati. è sempre necessario inserire self come primo elemento perchè è l'oggetto che stai inizializzando (si fa così e basta). Tutti i parametri (meno self) vanno messi all'interno di un attributo che si definisce con self. e poi il nome dell'attributo. Il metodo init viene chiamato metodo costruttore

METODO REPR

Il metodo __repr__ serve a definire una rappresentazione testuale di un oggetto: in pratica, dici a Python «quando vuoi mostrare questo oggetto in modo tecnico e informativo, usa questa stringa». È utile soprattutto per capire quali dati contiene un oggetto mentre programmi o cerchi un errore.



```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
    def __repr__(self):
        return f'{self.__class__.__name__} (titolo: {self.titolo}, artista: {self.artista}, anno: {self.anno}, durata: {self.durata}'

canzone=Canzone("someone like you", "Adele", "2011", "3.11")
print(canzone)
```

{self.__class__.__name__} serve a stapare "Canzone" prima della descrizione dei vari attributi

Tutte le canzoni ce vengono stampate anche nel main chiamando la classe Canzone che contiene il metodo repr saranno printate con la struttura definita nel repr.

METODO EQ

Se crei due oggetti canzone c1, c2 e poi nel main scrivi if c1==c2- Python confronta i riferimenti di c1 e c2 e quindi li darà sempre diversi

Per fare in modo di poter confrontare questi due oggetti devo definire nella classe il metodo __eq__

```python
class Canzone:
    def __init__(self, titolo, artista, anno, durata):
        self.titolo = titolo
        self.artista=artista
        self.anno=anno
        self.durata=durata
    def __eq__(self, other):
        if self.titolo==other.titolo:
            return True
        elif self.artista==other.artista:
            return True
        else:
            return False
```

in questo caso le due canzoni verronno considerate uguali se hanno lo stesso titolo o lo stesso autore
